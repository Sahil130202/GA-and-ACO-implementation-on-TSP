```python
import numpy as np
import pandas as pd
```

```python
num_rows = 10
data = {
    'Company': [f'Company_{i}' for i in range(num_rows)],
    'radio': np.random.randint(10, 100, num_rows),
    'sales': np.random.randint(1000, 10000, num_rows)
}
df = pd.DataFrame(data)

display(df.head())
```

```python
def predict_sales(radio,weights,bias):
  return weights*radio+bias
```

```python
def cost_function(radio,weights,sales, bias):
  companies = len(radio)
  total_error = 0.0
  for i in range(companies):
    total_error+=(sales[i]-(weights*radio[i]+bias))**2
  return total_error/companies
```

```python
def update_weights(radio,sales,weights,bias,learning_rate):
  weight_deriv=0
  bias_deriv=0
  companies = len(radio)
  for i in range(companies):
    weight_deriv += -2*radio[i]*(sales[i]-(weights*radio[i]+bias))
    bias_deriv += -2*(sales[i]-(weights*radio[i]+bias))
  new_weights = weights - (weight_deriv/companies)*learning_rate
  new_bias = bias - (bias_deriv/companies)*learning_rate
  return new_weights,new_bias
```

```python
def train(weights, bias, epochs, learning_rate, radio, sales):
  cost_history=[]
  for i in range(epochs):
    weights, bias= update_weights(radio,sales,weights,bias,learning_rate)
    cost=cost_function(radio,weights,sales,bias)
    cost_history.append(cost)

    if i % 10==0:
      print(f"Epoch {i} : {cost}")

  return weights, bias, cost_history
```

```python
weights = 0.0
bias = 0.0
learning_rate = 0.0001
epochs = 100

radio_data = df['radio'].values
sales_data = df['sales'].values

final_weights, final_bias, cost_history = train(weights, bias, epochs, learning_rate, radio_data, sales_data)

print(f"\nFinal Weights: {final_weights}")
print(f"Final Bias: {final_bias}")
```
